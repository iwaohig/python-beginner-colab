## 1. 比べる

2 つの値を比べると、結果は `True` (正しい) か `False` (まちがい) になります。

| 記号 | 意味 |
|---|---|
| `==` | 等しい (`=` が **2 つ**) |
| `!=` | 等しくない |
| `>` / `<` | 大きい / 小さい |
| `>=` / `<=` | 以上 / 以下 |

````python
temp = 18
print(temp > 20)
print(temp <= 20)
print(temp == 18)
print("晴れ" == "雨")
````

`=` (1 つ) は「箱に入れる」、`==` (2 つ) は「等しいか比べる」です。まちがえやすいので気をつけましょう。

## 2. もし○○なら (`if`)

```python
if 条件:
    条件が True のときにすること
```

- 条件のあとに **`:` (コロン)** をつける
- 次の行は **字下げ (インデント)** する。Colab では Enter を押すと自動で 2 つ空白が入る
- 字下げしている行が「`if` の中身」になる

````python
temp = 30

if temp >= 25:
    print("暑いので、水分をとりましょう")
print("この行は、いつでも表示されます")
````

`temp` を 15 に変えて、もう一度実行してみてください。1 行目のメッセージが出なくなります。

## 3. そうでなければ (`else`)・ほかの場合 (`elif`)

`elif` は「そうではなくて、もし○○なら」。上から順に調べて、**最初に当てはまったところだけ** 実行します。

````python
temp = 12

if temp >= 25:
    print("半袖で OK")
elif temp >= 15:
    print("長袖がちょうどいい")
else:
    print("上着を持っていこう")
````

## 4. 条件を組み合わせる

| 書き方 | 意味 |
|---|---|
| `A and B` | A と B の **両方** が True |
| `A or B` | A と B の **どちらか** が True |
| `not A` | A の反対 |

````python
weather = "雨"
temp = 10

if weather == "雨" and temp < 15:
    print("寒い雨の日です。暖かくしてね")

if weather == "雨" or weather == "雪":
    print("傘を持っていこう")
````

## やってみよう 1: 服装アドバイス

気温 `temp` に合わせて、次のように表示してください。

| 気温 | 表示 |
|---|---|
| 28 度以上 | 真夏の服装で。熱中症に注意 |
| 20 度以上 | 半袖で OK |
| 12 度以上 | 長袖か薄い上着を |
| それ未満 | コートを着ていこう |

`temp` をいろいろ変えて、どれも正しく出るか確かめましょう (28、20、19、12、5 など、**境目の数** を試すのがコツです)。

````python
temp = 19

if temp >= 28:
    print("真夏の服装で。熱中症に注意")
elif temp >= 20:
    print("半袖で OK")
elif temp >= 12:
    print("長袖か薄い上着を")
else:
    print("コートを着ていこう")
````

## やってみよう 2: 成績の判定

点数 `score` から評価を表示してください。

| 点数 | 評価 |
|---|---|
| 90 以上 | S |
| 80 以上 | A |
| 70 以上 | B |
| 60 以上 | C |
| それ未満 | D (不合格) |

````python
score = 76

if score >= 90:
    print("S")
elif score >= 80:
    print("A")
elif score >= 70:
    print("B")
elif score >= 60:
    print("C")
else:
    print("D (不合格)")
````

## やってみよう 3: 入場料

ある施設の入場料は、ふつう 1,200 円です。ただし、

- 12 歳以下は 500 円
- 学生 (`is_student` が `True`) は 800 円
- 65 歳以上は 600 円

年齢 `age` と学生かどうか `is_student` から、入場料を表示してください。
12 歳以下の学生は、安いほう (500 円) にします。

````python
age = 19
is_student = True

if age <= 12:
    fee = 500
elif age >= 65:
    fee = 600
elif is_student:
    fee = 800
else:
    fee = 1200
print(f"入場料は {fee} 円です")
````

## 発展: 偶数か奇数か

`%` (わり算の余り) を使うと、偶数か奇数かを調べられます。2 でわって余りが 0 なら偶数です。
`input` で数を受け取って、偶数か奇数かを表示してください。

````python
number = int(input("数を入れてください: "))

if number % 2 == 0:
    print(f"{number} は偶数です")
else:
    print(f"{number} は奇数です")
````
