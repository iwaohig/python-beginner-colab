## 1. リスト

**リスト** は、たくさんの値を順番に並べて入れておく入れものです。`[ ]` で囲み、`,` で区切ります。
第 1 回のおみくじで使ったのもリストです。

````python
fruits = ["りんご", "みかん", "ぶどう"]
print(fruits)
print(len(fruits))  # len は「いくつ入っているか」
````

### 中身を 1 つ取り出す

`リスト[番号]` で取り出せます。**番号は 0 から始まります** (1 番目が `[0]`)。
`[-1]` は「いちばん最後」です。

````python
fruits = ["りんご", "みかん", "ぶどう"]
print(fruits[0])
print(fruits[1])
print(fruits[-1])
````

### 追加する・入れ替える

````python
fruits = ["りんご", "みかん", "ぶどう"]
fruits.append("もも")  # 最後に追加
fruits[1] = "いちご"   # 2 番目を入れ替え
print(fruits)
````

## 2. くり返し (`for`)

```python
for 変数 in リスト:
    くり返すこと
```

リストの中身が 1 つずつ変数に入って、字下げした部分が中身の数だけくり返されます。
`if` と同じく、`:` と字下げが必要です。

````python
fruits = ["りんご", "みかん", "ぶどう"]

for fruit in fruits:
    print(f"{fruit}が好きです")
print("おしまい")
````

### 決まった回数くり返す (`range`)

`range(5)` は「0, 1, 2, 3, 4」の 5 つの数を作ります (5 は含みません)。
`range(1, 6)` なら「1, 2, 3, 4, 5」です。

````python
for i in range(1, 6):
    print(f"{i} 回目")
````

## 3. くり返しで合計を出す

「合計の箱を 0 にしておき、1 つずつ足していく」のが基本の形です。
`total += x` は `total = total + x` の短い書き方です。

````python
prices = [450, 300, 680, 120]

total = 0
for price in prices:
    total += price
print(f"合計 {total} 円")
````

よく使う計算は、最初から道具が用意されています。

````python
prices = [450, 300, 680, 120]

print(sum(prices))  # 合計
print(max(prices))  # いちばん大きい
print(min(prices))  # いちばん小さい
print(sum(prices) / len(prices))  # 平均
````

## 4. くり返し + 条件分岐

`for` の中に `if` を入れると、「条件に合うものだけ」を数えたり選んだりできます。
字下げが 2 段になることに注意してください。

````python
prices = [450, 300, 680, 120]

count = 0
for price in prices:
    if price >= 400:
        print(f"{price} 円は 400 円以上")
        count += 1
print(f"400 円以上は {count} 個")
````

## やってみよう 1: 1 週間の歩数

月曜から日曜までの歩数が `steps` に入っています。次のものを表示してください。

1. 1 週間の合計
2. 1 日の平均 (小数第 1 位まで。`round` を使う)
3. いちばん多く歩いた日の歩数
4. 8,000 歩以上歩いた日の数

````python
steps = [6500, 8200, 7400, 10300, 5600, 12100, 9000]

print(f"合計: {sum(steps)} 歩")
print(f"平均: {round(sum(steps) / len(steps), 1)} 歩")
print(f"最大: {max(steps)} 歩")

count = 0
for s in steps:
    if s >= 8000:
        count += 1
print(f"8,000 歩以上の日: {count} 日")
````

## やってみよう 2: 曜日といっしょに表示

曜日のリスト `days` と歩数のリスト `steps` は、同じ番号どうしが対応しています
(`days[0]` の歩数が `steps[0]`)。次のように 7 行表示してください。
8,000 歩以上の日には「目標達成!」をつけます。

```
月: 6500 歩
火: 8200 歩 目標達成!
...
```

ヒント: `for i in range(7):` として、`days[i]` と `steps[i]` を使う

````python
days = ["月", "火", "水", "木", "金", "土", "日"]
steps = [6500, 8200, 7400, 10300, 5600, 12100, 9000]

for i in range(7):
    if steps[i] >= 8000:
        print(f"{days[i]}: {steps[i]} 歩 目標達成!")
    else:
        print(f"{days[i]}: {steps[i]} 歩")
````

## やってみよう 3: 九九の表

`input` で 1〜9 の数を受け取り、その段の九九を表示してください。

```
7 × 1 = 7
7 × 2 = 14
...
7 × 9 = 63
```

````python
n = int(input("何の段? "))

for i in range(1, 10):
    print(f"{n} × {i} = {n * i}")
````

### 問題 3

エラーは出ませんが、合計がおかしくなります (正しくは 1,550 円)。なぜでしょう?

````python
prices = [450, 300, 680, 120]

total = 0  # 0 にするのは、くり返しの「前」に 1 回だけ
for price in prices:
    total += price
print(f"合計 {total} 円")
````
