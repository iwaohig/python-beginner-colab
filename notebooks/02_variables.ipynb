## 1. 変数は「名札のついた箱」

**変数** は、値に名前をつけて覚えておくしくみです。`=` は「等しい」ではなく、
**「右の値を、左の名前の箱に入れる」** という意味です。

````python
name = "山田 太郎"
print(name)
````

````python
price = 450
count = 3
print(price * count)
````

箱の中身は入れ替えられます。新しい値を入れると、前の値は消えます。

````python
weather = "晴れ"
print(weather)

weather = "雨"
print(weather)
````

### 変数の名前のつけかた

- 英字・数字・`_` (アンダースコア) を使う。数字から始めてはいけない
- 中身が分かる名前にする (`a` より `price` のほうが読みやすい)
- 大文字と小文字は区別される (`Price` と `price` は別の箱)

## 2. 数と文字

Python の値にはいくつか **型** (種類) があります。`type( )` で確かめられます。

| 型 | 意味 | 例 |
|---|---|---|
| `int` | 整数 | `20`, `-3` |
| `float` | 小数 | `1.5`, `36.8` |
| `str` | 文字列 | `"こんにちは"`, `"20"` |

````python
print(type(20))
print(type(36.8))
print(type("20"))
````

### 計算の記号 (前回の続き)

| 記号 | 意味 | 例 |
|---|---|---|
| `//` | わり算の商 (整数) | `7 // 2` → 3 |
| `%` | わり算の余り | `7 % 2` → 1 |
| `**` | べき乗 | `2 ** 3` → 8 |

````python
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(2 ** 10)
````

## 3. 文字をつなげる

文字列どうしは `+` でつなげられます。でも、**文字列と数はそのままではつなげられません**。
そういうときは **f 文字列** が便利です。`" "` の前に `f` をつけ、`{ }` の中に変数を書きます。

````python
name = "山田 太郎"
age = 19

print("こんにちは、" + name + "さん")
print(f"{name}さんは {age} 歳です")
print(f"来年は {age + 1} 歳です")
````

## 4. 入力を受け取る

`input( )` を使うと、実行したときにキーボードから入力できます。
セルを実行すると入力欄が出るので、文字を入れて Enter を押してください。

````python
name = input("お名前は? ")
print(f"{name}さん、ようこそ!")
````

**注意**: `input` で受け取った値は、数字を入れても **文字列** です。
計算に使うときは `int( )` (整数に) や `float( )` (小数に) で変換します。

````python
year_text = input("生まれた年は? (例: 2007) ")
year = int(year_text)
print(f"2026 年の誕生日で {2026 - year} 歳です")
````
