## 3. 計算もできる

数字は `" "` で囲まずに書きます。計算の記号は次のとおりです。

| 記号 | 意味 | 例 |
|---|---|---|
| `+` | たす | `3 + 2` → 5 |
| `-` | ひく | `3 - 2` → 1 |
| `*` | かける | `3 * 2` → 6 |
| `/` | わる | `3 / 2` → 1.5 |

```python
print(3 + 2)
print(3 * 2)
print(100 / 4)
```

`" "` で囲むと、計算ではなく「文字」として扱われます。違いを確かめてみましょう。

```python
print(1 + 2)
print("1 + 2")
```
