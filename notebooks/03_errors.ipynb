### エラーの読み方

1. **いちばん下の行を見る**。`○○Error: 説明` の形になっている
2. `○○Error` の名前で、だいたいの種類が分かる
3. 上のほうにある **矢印 (`^`) や行番号** で、どこがおかしいかを見る

```python
message = "おはよう"
print(mesage)
```

いちばん下の行は `NameError: name 'mesage' is not defined. Did you mean: 'message'?` です。
「`mesage` という名前はありません。もしかして `message`?」と教えてくれています。

## 2. よく出るエラー

| エラー | 意味 | よくある原因 |
|---|---|---|
| `SyntaxError` | 書き方がおかしい | かっこ・`" "` の閉じ忘れ、**全角の記号** |
| `NameError` | その名前がない | 綴りの間違い、まだ作っていない変数を使った |
| `TypeError` | 型が合わない | 文字列と数を `+` でつなげた |
| `IndentationError` | 字下げがおかしい | 行の先頭に余計な空白がある |
| `ZeroDivisionError` | 0 でわった | 割る数が 0 になっている |

### 日本語で入力するときの落とし穴: 全角

Python の記号は **すべて半角** です。日本語入力のまま `（` `）` `”` やスペースを打つと、
見た目はほとんど同じでもエラーになります。

```python
print（"全角のかっこ"）
```

`invalid character '（' (U+FF08)` は「使えない文字 `（` があります」という意味です。
困ったら、記号を一度消して **半角で打ち直す** のが早道です。

```python
print("値段は" + 300 + "円です")
```

`TypeError: can only concatenate str (not "int") to str` は
「文字列には文字列しかつなげられません (int はだめ)」という意味です。
`str(300)` で文字列に変えるか、f 文字列を使えば直ります。
